Finally, let's count the number of pixels where data are present in the L3 product for each day of the time series. This will indicate what percentage of the region is affected by missing data for each day.

We can transform an array so that each pixel takes a value of 1 if data is present and 0 if the data is missing by using a boolean comparison. This is similar to the approach that you used when applying thresholds yesterday.

Let's look at an example again:

```python
import numpy as np
# create an array with some missing points:
A = np.array([5, 6, np.nan, 8, np.nan])
print('A=',A)
# we have 3 data values, and 2 missing values

# we will want to transform this array into: [1, 1, 0, 1, 0]
# so the data points -> 1 in the new array, and the
# np.nan values -> 0

# let's use the boolean approach:
B = A == A
print('B=',B)

# like yesterday, we have a True/False array
# we can choose to convert this into 0s and 1s if we wish
# by using B.astype(int):
print('B.astype(int)=',B.astype(int))

# but to sum the elements this is not actually necessary:
# if we calculate a sum, True -> 1 and False -> 0:
print('np.sum(B)=',np.sum(B))
print('np.sum(B.astype(int))=',np.sum(B.astype(int)))

# you can see the results are the same
```
